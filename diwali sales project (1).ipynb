```python
df.drop(['Status','unnamed1'], axis=1,inplace=True)
```

```python
df.info()
```

#### Here we are checking the total gender of femal and male 

```python
ax = sns.countplot(x='Gender',data= df)
for bars in ax.containers:
    ax.bar_label(bars)
    
```

From above graps we can see that most of the buyers are females and even the purchasing power of females are greater then man

```python
sales_gen = df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Gender', y= 'Amount',data= sales_gen)
```

```python
df.columns
```

### analysing the age agroup 

```python
ax = sns.countplot(data=df, x = 'Age Group', hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

### Most of the amt spend as per age

```python
sales_gen = df.groupby(['Age Group'], as_index = False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Age Group', y= 'Amount',data= sales_gen,hue='Age Group')
```

### its a state based on highest orders 

```python
sales_state = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
sns.set(rc={'figure.figsize':(19,5)})
sns.barplot(data = sales_state, x='State', y= 'Orders', hue='State')
```

```python
df.columns
```

### analysing the marital_status

```python
ax = sns.countplot(data=df, x = 'Marital_Status', hue='Marital_Status')
sns.set(rc={'figure.figsize':(15,8)})
for bars in ax.containers:
    ax.bar_label(bars)
```

### checking the total count of male and female based on Marital_status 

```python
Marid_status = df.groupby(['Marital_Status','Gender'], as_index = False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(data = Marid_status, x='Marital_Status', y= 'Amount', hue='Gender')
```

### gathering the information baased on occupation 

```python
ax = sns.countplot(data=df, x = 'Occupation', hue='Occupation')
sns.set(rc={'figure.figsize':(25,15)})
for bars in ax.containers:
    ax.bar_label(bars)
```

### Analysing the product_category distrubition count based on feature values.

```python
ax = sns.countplot(data=df, x = 'Product_Category', hue='Product_Category')
sns.set(rc={'figure.figsize':(25,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

## Analysing the amount spending over the product
 --> we used only 10 headrs value.

```python
product =df.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by= 'Amount', ascending= False).head(25)
sns.set(rc={'figure.figsize':(35,10)})
sns.barplot(data = product, x='Product_Category', y= 'Amount', hue='Product_Category')
```

### how much spemding over the diferent Zone by amount.

```python

plt.figure(figsize=(15, 5))

for zone in df['Zone'].unique():
    subset = df[df['Zone'] == zone]
    sns.kdeplot(data=subset, x='Amount', label=zone, fill=False)

plt.legend(title='Zone')
plt.title('KDE of Amount by Zone')
plt.show()

```
